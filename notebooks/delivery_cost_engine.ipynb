```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import gamma

print("📊 Compiling Ingestion Feature Blueprints: Trip Distance Profile...")

# 1. Coordinate Grid Initialization
x_distance = np.linspace(0, 15, 1000)

# 2. Calculate the exact Gamma Probability Density Function (PDF)
# Shape=2, Scale=1.5 (Shifted by +1 to enforce a realistic 1-mile minimum distance floor)
y_gamma = gamma.pdf(x_distance - 1, a=2, scale=1.5)

# 3. Canvas Composition
fig, ax = plt.subplots(figsize=(6, 4))

ax.plot(x_distance, y_gamma, color="#c62828", lw=3, label="Trip Distance Profile")
ax.fill_between(x_distance, y_gamma, color="#c62828", alpha=0.15)

# Enforce our hard structural floor constraint visually
ax.axvline(1.0, color="#212121", linestyle=":", lw=2, label="Hard Minimum Floor (1.0 Mile)")

# Visual highlights for complete amateur formatting clarity
ax.set_title("Feature 1: Trip Distance Distribution\n(Right-Skewed Gamma Profile)", fontsize=11, fontweight="bold")
ax.set_xlabel("Distance Traveled (Miles)", fontsize=9)
ax.set_ylabel("Probability Vector Density", fontsize=9)
ax.set_xlim(0, 12)
ax.set_ylim(0, 0.3)
ax.grid(True, linestyle=":", alpha=0.5)
ax.legend(fontsize=9, loc="upper right")

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import beta

print("📊 Compiling Ingestion Feature Blueprints: Driver Availability Profile...")

# 1. Coordinate Grid Initialization between absolute boundaries
x_availability = np.linspace(0.001, 0.999, 1000)

# 2. Calculate the exact U-Shaped Beta Probability Density Function (PDF)
# Alpha=0.5, Beta=0.5 forces the curve to crater in the center and rocket at the edges
y_beta = beta.pdf(x_availability, a=0.5, b=0.5)

# 3. Canvas Composition
fig, ax = plt.subplots(figsize=(6, 4))

ax.plot(x_availability, y_beta, color="#2e7d32", lw=3, label="Driver Availability Profile")
ax.fill_between(x_availability, y_beta, color="#2e7d32", alpha=0.15)

# Visual anchors highlighting the boundary constraints
ax.set_title("Feature 2: Driver Availability Ratio\n(U-Shaped Bounded Beta Profile)", fontsize=11, fontweight="bold")
ax.set_xlabel("Availability Ratio (0.0 = Critical Shortage, 1.0 = Extreme Surplus)", fontsize=9)
ax.set_ylabel("Probability Vector Density", fontsize=9)
ax.set_xlim(0, 1)
ax.set_ylim(0, 3.5)
ax.grid(True, linestyle=":", alpha=0.5)
ax.legend(fontsize=9, loc="upper center")

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

print("🛰️ Synthesizing 3D Delivery Cost Interaction Topography...")

# 1. Create a coordinate grid (Chessboard) matching our two moving market forces
distance_axes = np.linspace(1.0, 12.0, 100)      # Trip distance from 1 to 12 miles
availability_axes = np.linspace(0.01, 0.99, 100) # Availability ratio from 0.01 to 0.99
Distance_Grid, Availability_Grid = np.meshgrid(distance_axes, availability_axes)

# 2. Model the real-world business logic: Non-Linear Compounding Interaction Formula
# Base price scales with distance, then explodes non-linearly at critical driver shortages (near 0.0)
base_distance_cost = Distance_Grid * 1.5
shortage_pressure_multiplier = 1.0 / (Availability_Grid + 0.05)
Delivery_Cost_Surface = base_distance_cost * shortage_pressure_multiplier

# 3. Assemble and display the interactive 3D Topography Chart
fig = plt.figure(figsize=(8, 5.5))
ax = fig.add_subplot(111, projection='3d')

surf = ax.plot_surface(Distance_Grid, Availability_Grid, Delivery_Cost_Surface, 
                       cmap='turbo', edgecolor='none', alpha=0.9)

# Formatting anchors for complete layman scannability
ax.set_title("Delivery Optimization Topography: How Distance and Shortages Compound", 
             fontsize=11, fontweight="bold", pad=15)
ax.set_xlabel("Trip Distance (Miles)", fontsize=8, labelpad=8)
ax.set_ylabel("Driver Availability Ratio", fontsize=8, labelpad=8)
ax.set_zlabel("Final Delivery Cost ($)", fontsize=8, labelpad=8)

# Spin the 3D angle view for perfect structural perspective
ax.view_init(elev=22, azim=-140)

# Add clear color bar tracking the pricing risk scale
cbar = fig.colorbar(surf, ax=ax, shrink=0.5, aspect=10, pad=0.1)
cbar.set_label("Price Surge Multiplier Intensity Scale", fontsize=8, fontweight="bold")

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

print("⚙️ Compiling Continuous Regression Loss Metric Penalty Profiles...")

# 1. Simulate a continuous range of absolute delivery price errors (residuals) up to $0.25
delivery_price_errors = np.linspace(0, 0.25, 500)

# 2. Calculate the corresponding penalty weights under linear (MAE) vs quadratic (RMSE) metrics
mae_linear_penalty = delivery_price_errors
rmse_quadratic_penalty = delivery_price_errors ** 2  # The squaring effect

# 3. Assemble the corporate-grade metric comparison chart
fig, ax = plt.subplots(figsize=(7, 4))

ax.plot(delivery_price_errors, mae_linear_penalty, color="#0288d1", lw=2.5, 
        label="MAE Penalty Profile (Linear Scaling: e)")
ax.plot(delivery_price_errors, rmse_quadratic_penalty, color="#d32f2f", lw=2.5, linestyle="--", 
        label="RMSE Penalty Profile (Quadratic Squaring: e²)")

# Shade the high-risk zone where large errors are heavily punished
ax.fill_between(delivery_price_errors, rmse_quadratic_penalty, color="#d32f2f", alpha=0.08)

# Add instructional annotation anchors
ax.annotate('Catastrophic Error Zone\n(RMSE Punishes Blunders)', xy=(0.22, 0.05), xytext=(0.06, 0.16),
            arrowprops=dict(arrowstyle="->", lw=1.5, color="#333333"), fontsize=9, fontweight='bold')

ax.set_title("E-Commerce Loss Metrics: MAE vs. RMSE Penalty Scaling", fontsize=11, fontweight="bold")
ax.set_xlabel("Absolute Prediction Error Magnitude ($)", fontsize=9)
ax.set_ylabel("Optimization Cost Penalty Weight", fontsize=9)
ax.grid(True, linestyle=":", alpha=0.6)
ax.legend(loc="upper left", fontsize=9)

plt.tight_layout()
plt.show()
```
